```python
from pyspark.sql import functions as F

source_path = "/Volumes/workspace/default/book_data/first_project_orders.csv"

raw_orders_df = (
    spark.read
    .option("header", "true")
    .csv(source_path)
)

bronze_df = (
    raw_orders_df
    .withColumn("source_file", F.col("_metadata.file_path"))
    .withColumn("loaded_at", F.current_timestamp())
)

bronze_df.write.mode("overwrite").saveAsTable(
    "workspace.first_project.orders_bronze"
)

print(
    "bronze rows:",
    spark.table("workspace.first_project.orders_bronze").count()
)

spark.table("workspace.first_project.orders_bronze").show(5)
```
